```python
from ffox_test import FirefoxSession

# RFB opens the URL; BiDi only joins for DOM work.
async with FirefoxSession() as ff:
    await ff.navigate(
        "https://the-internet.herokuapp.com/login",
        ready="#username",
    )
    await ff.fill("#username", "tomsmith")
    await ff.fill("#password", "SuperSecretPassword!")
    await ff.click("button[type=submit]")
    await ff.wait_for_selector(".flash.success")
    print(await ff.current_url())
    print(await ff.text_content(".flash.success"))
```
